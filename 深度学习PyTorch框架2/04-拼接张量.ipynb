```python
import torch
```

```python
"""
总结:cat:1.不能修改张量的维度个数。例如：不能将2维改为3维
        2.除了拼接的维度之外，其他的维度都必须相同
"""
```

```python

t1 = torch.randint(1,10,size=(2,3))
t2 = torch.randint(1,10,size=(2,3))
print(t1,t1.shape)
print(t2,t2.shape)
```

```python
#将t1 和 t2 在第0个维度上拼接起来，得到一个新的张量cat_1
#注意：按照行拼接，则列数要相同，
t3 = torch.cat([t1,t2],dim=0)
print(t3,t3.shape)
t4 = torch.cat([t1,t2],dim=1)
print(t4,t4.shape)
```

```python
"""
    cat：
        1- 不能修改张量的维度个数。例如：不能将2维变3维
        2- 除了拼接的维度以外，其他维度必须相同
"""
t1 = torch.randint(1,10,size=(2,3))
t2 = torch.randint(1,10,size=(2,3))
print(t1,t1.shape)
print(t2,t2.shape)
```

```python
# 需求：将t1和t2在第0个维度上拼接起来, 得到一个新的张量cat_1
# dim=0：拼接在第0个维度上
cat_1 = torch.cat([t1,t2],dim=0)   # [4,3] cat_1.shape = [4,3]
print(cat_1,cat_1.shape)

# dim=1：拼接在第1个维度上
cat_2 = torch.cat([t1,t2],dim=1)
print(cat_2,cat_2.shape)

# TODO 注意：按照行拼接，要求列数相同，按照列拼接，要求行数相同
# 不能将2维变3维
# torch.cat([t1,t2],dim=2)
```

```python
t1 = torch.randint(1,10,size=(2,3))
# 除了拼接的维度以外，其他维度必须相同
t2 = torch.randint(1,10,size=(5,3))
# t2 = torch.randint(1,10,size=(2,4))
print(t1,t1.shape)
print(t2,t2.shape)

cat_1 = torch.cat([t1,t2],dim=0)
print(cat_1,cat_1.shape)
```

```python
"""
    stack：
        1- 两个拼接的张量形状必须完全一样
        2- 会产生新维度，在新维度上进行拼接操作
"""
t1 = torch.randint(1,10,size=(5,6))
t2 = torch.randint(1,10,size=(5,6))
print(t1,t1.shape)
print(t2,t2.shape)

stack_1 = torch.stack([t1,t2],dim=0)
print(stack_1,stack_1.shape) # [2,5,6]

stack_2 = torch.stack([t1,t2],dim=1)
print(stack_2, stack_2.shape) # [5,2,6]

print("---"*20)
stack_3 = torch.stack([t1,t2],dim=2)
print(stack_3, stack_3.shape) # [5,6,2]
```
