```python
#导入相关包
import torch
```

```python
#1.创建指定形状的全0张量
data = torch.zeros(2,3)
data
```

```python
#2.根据张量形状创建全0张量
a = torch.Tensor(2,3)
a
data = torch.zeros_like(a)
data
```

```python
#1.创建指定形状全1张量
data = torch.ones(2,3)
data
```

```python
#2.根据张量形状创建全1张量
data = torch.ones_like(data)
data
```

```python
#1.船舰指定形状指定值的张量
data = torch.full([2,3],10)
data
```

```python
#2.根据张量形状创建指定值的张量
data = torch.full_like(data,2)
data
```
