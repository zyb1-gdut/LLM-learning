```python
#导入相关包
import torch
```

```python
data = torch.randint(0,10,[2,3])
print(data)
#1.不修改原数据
new_data = data.add(10)#等价 new_data = data +10
print(new_data)
print(data )
#2.直接修改原数据 注意：带下划线的函数为需修改原数据本身
data.add_(10)
print(data)
#其他函数
print(data.sub(100))
print(data.mul(100))
print(data.div(100))
print(data.neg())
```

```python
#点乘运算
data1 = torch.tensor([[1,2],[3,4]])
data2 = torch.tensor([[5,6],[7,8]])
print(data1)
print(data2)
#第一种方式
data  = torch.mul(data1,data2)
print(data)
#第二种方式
data = data1 * data2
print(data)
```

```python
#矩阵乘法运算/点积
data1 = torch.tensor([[1,2],[3,4],[5,6]])
data2 = torch.tensor([[5,6],[7,8]])
print(data1)
print(data2)
#方式一
data3 = data1 @ data2
print('data3-->',data3)
#方式二
data4 = torch.matmul(data1,data2)
print('data4-->',data4)
```
