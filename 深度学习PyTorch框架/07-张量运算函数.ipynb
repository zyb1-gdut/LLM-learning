```python
#导入相关包
import torch
from conda_build.build import print_build_intermediate_warning
from langchain_core.utils import pre_init
```

```python
#常见函数
data = torch.randint(0,10,[2,3],dtype=torch.float64)
print(data)
```

```python
#1.计算均值
#注意：tensor必须为Float或者Double类型
print(data.mean())
print(data.mean(dim=0))#按列计算均值
print(data.mean(dim=1))#按行计算均值
```

```python
#2.计算总和
print(data.sum())
print(data.sum(dim=0))#按列
print(data.sum(dim=1))#按行
```

```python
#3.计算平方
print(torch.pow(data,2))
```

```python
#4.计算平方根
print(data.sqrt())
```

```python
#5.指数运算，e~n次方
print(data.exp())#e^6,e^2,e^0
```

```python
#6.对数运算
print(data.log())#以e为底
print(data.log2())
print(data.log10())
```
