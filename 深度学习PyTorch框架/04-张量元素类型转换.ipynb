```python
import torch
from numpy import dtype
```

```python
data = torch.full([2,3],10)
print(data,dtype)
```

```python
#将data元素类型转换为float64类型
data = data.type(torch.DoubleTensor)
print(data,dtype)
```

```python
#转为其他类型
data = data.type(torch.ShortTensor)
print(data,dtype)
```

```python
data = data.type(torch.IntTensor)
print(data,dtype)
```

```python
data = data.type(torch.LongTensor)
print(data,dtype)
```

```python
data = data.type(torch.FloatTensor)
print(data,dtype)
```

```python
data = data.type(dtype=torch.float16)
print(data,dtype)
```

```python
data = torch.full([2, 3], 10)
print (data.dtype)
#将 data 元素类型转换为 float64类型
data = data.double()
print (data.dtype)
```

```python
#转换为其他类型
data = data.short()
print (data.dtype)
```

```python
data = data.int()
print (data.dtype)
```

```python
data = data.long()
print (data.dtype)
```

```python
data = data.float()
print (data.dtype)
```
